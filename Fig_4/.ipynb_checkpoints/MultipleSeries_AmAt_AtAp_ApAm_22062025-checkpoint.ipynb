```python
df_dict_growth = {}

# Loop over series numbers and load each CSV
new_series_number = 0
replicate_number = 0
for replicate in replicates:
    base_path = fr'C:\Users\Redaelli\Desktop\PostDoc\ImageAnalysis\WithMidap\{experiment}\{replicate}'
    Series_list = [d for d in glob.glob(base_path + r"\*") if os.path.isdir(d)]
    df_per_series = []
    for series in Series_list:
        
        match = re.search(r"Series_(\d+)", series)
        if not match:
            continue
        series_number = int(match.group(1))
        try:
            analysis_dir = [d for d in os.listdir(os.path.join(series, 'raw_images_half')) if d.startswith('Analysis')][0]
            df = pd.read_csv(os.path.join(series, 'raw_images_half', analysis_dir, 'Growth_df.csv'))
        except:
            continue
        columns_to_drop = ['ID', 'Area', 'Centroid_x_diff', 'Centroid_y_diff', 'Area_change', 'distance']
        df = df.drop(columns=columns_to_drop)
        df['Species_names']  = df['Species'].map(code_name_species_mapping)
        df['Time'] = df.Timepoint.multiply(1/6)
        df['Series'] = new_series_number
        df['Replicate'] = replicate_number
        df_per_series.append(df)
        new_series_number += 1
    replicate_number += 2
    df_dict_growth[replicate] = pd.concat(df_per_series, ignore_index=True)
```

# Growth

```python
# Create list to collect processed data from each series
grouped_species_count = []

for key, df in df_dict_growth.items():
    
    # Count total cells per timepoint
    total_cells = df.groupby('Time').Mask_nb.nunique()
    df = df[df['Species'] < 3].copy() # exclude non recognised cells
    df_growing = df[df['Elongation_rate'] > 0].copy() # exclude non growing cells
    total_growing_cells = df_growing.groupby('Time').Mask_nb.nunique()
 
    # Count cells per species per timepoint
    species_counts = df_growing.groupby(['Time', 'Species']).Mask_nb.nunique().reset_index(name='GrowingCellsPerSpecies')
    
    # Merge with total to compute percentage
    species_counts['TotalGrowing'] = species_counts['Time'].map(total_growing_cells)
    species_counts['PercentagePerSpecies'] = (species_counts['GrowingCellsPerSpecies'] / species_counts['TotalGrowing']) * 100
 
    # % of all growing cells out of all cells at each time
    species_counts['TotalCells'] = species_counts['Time'].map(total_cells)
    species_counts['TotalGrowingPercentage'] = (species_counts['TotalGrowing'] / species_counts['TotalCells']) * 100  

    # Assign series column
    species_counts['Series'] = key
    grouped_species_count.append(species_counts)

# Concatenate all and average

df_growth_stat = pd.concat(grouped_species_count, ignore_index=True)

# Group and pivot data to get mean PercentagePerSpecies per Time and Species
df_avg_growth = df_growth_stat.groupby(['Time', 'Species'])['PercentagePerSpecies'].median().reset_index()
print(df_avg_growth)
# Pivot to wide format for stacking
df_pivot_growth = df_avg_growth.pivot(index='Time', columns='Species', values='PercentagePerSpecies')

percentage_total = df_growth_stat.groupby(['Time'])['TotalGrowingPercentage'].median().reset_index()

# Plotting
fig, ax1 = plt.subplots(figsize=FIGSIZE)

# Custom color map
custom_cmap = mcolors.LinearSegmentedColormap.from_list("custom_cmap", [COLOR_RED, COLOR_YELLOW, COLOR_BLUE])

# Bar plot
bars = df_pivot_growth[[0, 1, 2]].plot(kind='bar', stacked=True, color=[code_colors.get(col, 'gray') for col in df_pivot_growth.columns], ax=ax1, width=1.0, edgecolor='black', alpha=0.5, legend=False)

# Y-axis
ax1.set_ylabel('Species %', fontsize=FONTSIZE)
ax1.set_xlabel('Time [hr]', fontsize=FONTSIZE)

# X-ticks setup
x_values = percentage_total['Time']
tick_indices = np.arange(0, max(percentage_total.index), 10)
tick_labels = np.round(percentage_total['Time'].iloc[tick_indices], 0).astype(int)
ax1.set_xticks(tick_indices)
ax1.set_xticklabels(tick_labels, fontsize=FONTSIZE, rotation=90)
ax1.tick_params(axis='y', labelsize=FONTSIZE)
ax1.set_ylim([0, 100])  # adjust as needed
ax1.set_xlim([0, 10])

# Line plot: number of growing cells (right y-axis)
ax2 = ax1.twinx()
ax2.plot(percentage_total.index, percentage_total.TotalGrowingPercentage, color='black', marker='o', linestyle='-', linewidth=2)
ax2.set_ylabel('Growing cells %', color='black', fontsize=FONTSIZE)
ax2.tick_params(axis='y', labelcolor='black', labelsize=FONTSIZE)
ax2.set_ylim([0, 100])  # adjust as needed
ax2.set_xticks(tick_indices)
ax2.set_xticklabels(tick_labels, fontsize=22, rotation=90)


# Final layout
plt.tight_layout()
#plt.savefig(f'{output_directory}/Growth_species_time_series_avg.png', dpi=300)
plt.show()
```

# Elongation

```python
grouped_species_elongation = []

for key, df in df_dict_growth.items():
    # Filter valid cells
    df = df[df['Species'] < 3].copy()  # Keep recognized species
    df_growing = df[df['Elongation_rate'] > 0].copy()  # Only growing cells

    # Compute elongation contribution: Elongation_rate * 1 (or use another weight if available)

    # Total elongation per time
    total_elongation = df_growing.groupby('Time')['Elongation_rate'].sum()
   
    # Elongation per species per time
    species_elongation = df_growing.groupby(['Time', 'Species'])['Elongation_rate'].sum().reset_index(name='ElongationPerSpecies')
   
    # Add total elongation for % calculation
    species_elongation['TotalElongation'] = species_elongation['Time'].map(total_elongation)
    
    # Calculate Elongation % per species per time
    species_elongation['ElongationPercentagePerSpecies'] = (species_elongation['ElongationPerSpecies'] / species_elongation['TotalElongation']) * 100

    # Add Series identifier
    species_elongation['Series'] = key

    grouped_species_elongation.append(species_elongation)
    

# Combine all Series into one DataFrame
df_elongation = pd.concat(grouped_species_elongation, ignore_index=True)

# Optional: Average across Series if needed
df_avg_el = df_elongation.groupby(['Time', 'Species'])['ElongationPercentagePerSpecies'].median().reset_index()

# Pivot for stacked bar plotting
df_pivot_el = df_avg_el.pivot(index='Time', columns='Species', values='ElongationPercentagePerSpecies')

# Plotting
fig, ax1 = plt.subplots(figsize=FIGSIZE)

# Custom color map
custom_cmap = mcolors.LinearSegmentedColormap.from_list("custom_cmap", [COLOR_RED, COLOR_YELLOW, COLOR_BLUE])

# Bar plot
bars =df_pivot_el[[0, 1, 2]].plot(kind='bar', stacked=True, color=[code_colors.get(col, 'gray') for col in df_pivot_el.columns], ax=ax1, width=1.0, edgecolor='black', alpha=0.5, legend=False)

# Step 5: Customize ticks
xticks = ax1.get_xticks()
xticklabels = np.round(df_pivot_el.index,0).astype(str)

# Show only 1 tick every 5
ax1.set_xticks(xticks[::5])
ax1.set_xticklabels(xticklabels[::5])

# Labels & title
ax1.set_ylabel('Median Elongation %', fontsize = FONTSIZE)
ax1.set_xlabel('Time [hr]', fontsize = FONTSIZE)
ax1.set_title('Median Elongation % Species', fontsize = FONTSIZE)
ax1.legend(title='Species', bbox_to_anchor=(1.05, 1), loc='upper left')
ax1.grid(axis='y', linestyle='--', alpha=0.6)
plt.tight_layout()
#plt.savefig(f'./{output_directory}/Elongation_in_time_per_series.png')
plt.show()
```

# Biomass

```python
grouped_species_biomass = []

for key, df in df_dict_growth.items():
    df = df[df['Species'] < 3].copy()  # Keep recognized species
    df_growing = df[df['Elongation_rate'] > 0].copy()  # Only growing cells
    
    # Count number of cells per species per time
    cell_counts_per_species = df_growing.groupby(['Time', 'Species'])['Mask_nb'].nunique().reset_index(name='NumCellsPerSpecies')
    total_cell = df_growing.groupby('Time').Mask_nb.nunique()
    
    # Mean elongation rate per species per time
    mean_elongation = df_growing.groupby(['Time', 'Species'])['Elongation_rate'].mean().reset_index(name='MeanElongation')

    # Merge elongation and cell counts
    biomass_df = pd.merge(cell_counts_per_species, mean_elongation, on=['Time', 'Species'])

    # Compute biomass = mean elongation × number of cells
    biomass_df['BiomassPerSpecies'] = biomass_df['MeanElongation'] * biomass_df['NumCellsPerSpecies']
    
    # Total number of cells
    biomass_df['Numcells'] = biomass_df['Time'].map(total_cell)
    
    # Total biomass per time
    total_biomass = biomass_df.groupby('Time')['BiomassPerSpecies'].sum()
    
    # Map to get total per time and compute %
    biomass_df['TotalBiomass'] = biomass_df['Time'].map(total_biomass)
    biomass_df['BiomassPercentagePerSpecies'] = (biomass_df['BiomassPerSpecies'] / biomass_df['TotalBiomass']) * 100

    # Average elongation rate of the population
    average_elongation = df_growing.groupby('Time').Elongation_rate.mean()
    biomass_df['AverageElongation'] =  biomass_df['Time'].map(average_elongation)
    
    # Add series ID
    biomass_df['Series'] = key
    
    grouped_species_biomass.append(biomass_df)

# Combine all series
df_biomass = pd.concat(grouped_species_biomass, ignore_index=True)

# Optional: average across Series
df_avg_biomass = df_biomass.groupby(['Time', 'Species'])['BiomassPercentagePerSpecies'].median().reset_index()

# Pivot for plotting
df_pivot_biomass = df_avg_biomass.pivot(index='Time', columns='Species', values='BiomassPercentagePerSpecies')

# Plotting
fig, ax1 = plt.subplots(figsize=FIGSIZE)

# Custom color map
custom_cmap = mcolors.LinearSegmentedColormap.from_list("custom_cmap", [COLOR_RED, COLOR_YELLOW, COLOR_BLUE])

# Bar plot
ax = df_pivot_biomass[[0, 1, 2]].plot(kind='bar', stacked=True, color=[code_colors.get(col, 'gray') for col in df_pivot_biomass.columns], ax=ax1, width=1.0, edgecolor='black', alpha=0.5, legend=False)


# Step 5: Customize ticks
xticks = ax.get_xticks()
xticklabels = np.round(df_pivot_biomass.index,0).astype(str)

# Show only 1 tick every 5
ax.set_xticks(xticks[::5])
ax.set_xticklabels(xticklabels[::5])

# Labels & title
ax.set_ylabel('Median Biomass Percentage', fontsize = FONTSIZE)
ax.set_xlabel('Time [hr]', fontsize = FONTSIZE)
#ax.set_title('Elongation Percentage Over Time by Species', fontsize = FONTSIZE)
ax.grid(axis='y', linestyle='--', alpha=0.6)

# Compute average (or median) TotalBiomass over time for the line plot
biomass_over_time = df_biomass.groupby('Time')['TotalBiomass'].mean()
elongaton_rate_over_time = df_biomass.groupby('Time')['AverageElongation'].median()

# Create a second y-axis sharing the same x-axis
ax2 = ax.twinx()

# Plot TotalBiomass on the secondary axis
#ax2.plot(np.arange(len(df_pivot_biomass)), biomass_over_time.values, color='lightgreen', marker='o', linestyle='-', linewidth=2, label='Total Biomass')
ax2.plot(np.arange(len(df_pivot_biomass)), elongaton_rate_over_time.values, color='black', marker='o', linestyle='-', linewidth=2, label='Average Elongation rate')

# Label for secondary y-axis
ax2.set_ylabel('Average El rate  [um / h]', color='black', fontsize = FONTSIZE)

# Optional: align tick colors
ax2.tick_params(axis='y', labelcolor='black')
ax2.set_ylim([0.05, 1.1*max(elongaton_rate_over_time.values)])
# Add second legend
lines1, labels1 = ax1.get_legend_handles_labels()
lines2, labels2 = ax2.get_legend_handles_labels()
ax2.legend(lines1 + lines2, labels1 + labels2)


plt.tight_layout()
plt.savefig(f'./{output_directory}/Biomass_in_time_per_series.png')
plt.show()
```

# Growth - Mixing extrapolation

```python
def LoadStackData(folder):
    """
    Load feature tables for any frame and concatenate them adding timepoint information.
    """
    files = glob.glob(os.path.join(folder, "analysis_*.csv"))
    files = sorted(files, key=lambda x: int(x.split('time')[-1].split('.csv')[0]))
    dataframes = []
    for file in files:
        timepoint = int(file.split("time")[-1].split(".csv")[0])  # Extract timepoint from filename
        df = pd.read_csv(file)
        density_columns = [col for col in df.columns if col.startswith('Local_density_r_')]
        unidentified_columns = [col for col in df.columns if col.startswith('Unidentified_pc_r_')]
        columns_to_drop = (['Timepoint', 'Centroid_y', 'Centroid_x', 'area', 'axis_major_length', 'Mother_mask',  'centroid-0', 'centroid-1']
                    + density_columns
                    + unidentified_columns)
        df = df.drop(columns=columns_to_drop)
        columns_to_fill = [col for col in df.columns if col.startswith('Occ_spec_')]
        df[columns_to_fill] = df[columns_to_fill].fillna(0.0)
        df['Timepoint'] = timepoint
        df['Species_names']  = df['Species'].map(code_name_species_mapping)
        dataframes.append(df)
    return pd.concat(dataframes, ignore_index=True)
```

```python
df_dict_mixing = {}
# Loop over series numbers and load each CSV
for series in Series_list:
    file_path = f'C:/Users/Redaelli/Desktop/PostDoc/ImageAnalysis/WithMidap/{experiment}/Series_{series}/raw_images_half/Analysis_rescaled_mixing/'
    df_mixing = LoadStackData(file_path)
    df_dict_mixing[series] = df_mixing
```

```python
# Concatenate growth and mixing dictionaries

columns_to_drop = ['Unnamed: 0', 'ID', 'centroid-0', 'centroid-1', 'Species', 'Species_names', 'axis_major_length', 'Timepoint', 'Area', 'Centroid_x_diff', 'Centroid_y_diff', 'Area_change', 'distance', 'centroid-0', 'centroid-1']

# Drop unwanted columns
df_dict_growth_clean_column = {}
for key in df_dict_growth:
    df_dict_growth_clean_column[key] = df_dict_growth[key].drop(columns=columns_to_drop, errors='ignore')

# Drop non-growing or unjustified growth
for key in df_dict_growth:
    df = df_dict_growth_clean_column[key].copy()
    df_dict_growth_clean_column[key] =  df[(df['Elongation_rate'] > 0) & (df['Elongation_rate'] < 1.5)]

# merge with mixing
df_dict_growth_mixing = {
    key: pd.merge(df_dict_growth_clean_column[key], df_dict_mixing[key], on='Mask_nb')
    for key in df_dict_growth
}

# Clean from 0 mixing values
df_dict_growth_mixing = {
    k: v[v[[col for col in v.columns if 'Mixing_rel' in col]].ge(0).all(axis=1)]
    for k, v in df_dict_growth_mixing.items()
}
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Prepare figure with 3 3D subplots (1 row, 3 columns)
fig, axes = plt.subplots(1, 3, figsize=(18, 6), subplot_kw={'projection': '3d'})

for i, (species_code, species_name) in enumerate(code_name_species_mapping.items()):
    ax = axes[i]

    species_closeby_1 = (species_code + 1) % 3
    species_closeby_2 = (species_code + 2) % 3

    color_ = code_colors[species_code]

    conc_species_1 = []
    conc_species_2 = []
    growth_species_focal = []

    for key, df_mixing in df_dict_growth_mixing.items():
        df_species = df_mixing[df_mixing['Species'] == species_code]

        conc_species_1.append(df_species[f'Occ_spec_{species_closeby_1}.0_r_76'].values)
        conc_species_2.append(df_species[f'Occ_spec_{species_closeby_2}.0_r_76'].values)
        growth_species_focal.append(df_species['Elongation_rate'].values)

    x = np.concatenate(conc_species_1)
    y = np.concatenate(conc_species_2)
    z = np.concatenate(growth_species_focal)

    ax.scatter(x, y, z, color=color_, alpha=0.6, s=20)

    ax.set_xlabel(f"Occupation {code_name_species_mapping[species_closeby_1]}", fontsize=12)
    ax.set_ylabel(f"Occupation {code_name_species_mapping[species_closeby_2]}", fontsize=12)
    ax.set_zlabel('Growth (Elongation rate)', fontsize=12)
    ax.set_title(f"{species_name}", fontsize=14)
    ax.grid(True)

plt.tight_layout()
plt.show()
```

```python
import statsmodels.api as sm

# Prepare figure for bar plots: one subplot per species
fig, axes = plt.subplots(3, 1, figsize=(5, 15))

for i, (species_code, species_name) in enumerate(code_name_species_mapping.items()):
    ax = axes[i]
    
    species_closeby_1 = (species_code + 1) % 3
    species_closeby_2 = (species_code + 2) % 3
    color_ = code_colors[species_code]
    conc_species_1 = []
    conc_species_2 = []
    growth_species_focal = []

    # Collect data from all datasets
    for key, df_mixing in df_dict_growth_mixing.items():
        df_species = df_mixing[df_mixing['Species'] == species_code]

        conc_species_1.append(df_species[f'Occ_spec_{species_closeby_1}.0_r_153'].values)
        conc_species_2.append(df_species[f'Occ_spec_{species_closeby_2}.0_r_153'].values)
        growth_species_focal.append(df_species['Elongation_rate'].values)

    x1 = np.concatenate(conc_species_1)
    x1 = np.nan_to_num(x1, nan=0.0, posinf=0.0, neginf=0.0)
    x2 = np.concatenate(conc_species_2)
    x2 = np.nan_to_num(x2, nan=0.0, posinf=0.0, neginf=0.0)
    y = np.concatenate(growth_species_focal)

    # Prepare design matrix with intercept
    X = np.column_stack((x1, x2))
    X = sm.add_constant(X)
    
    # Fit linear regression model
    model = sm.OLS(y, X).fit()

    # Extract coefficients (const, x1, x2)
    coef_names = ['Intercept', f'{code_name_species_mapping[species_closeby_1]}', f'{code_name_species_mapping[species_closeby_2]}']
    coefs = model.params
    errors = model.bse  # standard errors for error bars
    
    # Bar plot of coefficients with error bars
    ax.bar(coef_names, coefs, yerr=errors, capsize=5, color=color_)
    ax.set_title(f'{species_name}', fontsize=FONTSIZE)
    ax.set_ylabel('Coefficient', fontsize=FONTSIZE)
    ax.set_ylim(min(np.min(coefs)*1.2, 0), max(np.max(coefs)*1.2,0))
    ax.tick_params(axis='x', labelsize=FONTSIZE)  # Replace FONTSIZE with your desired size if needed
    ax.tick_params(axis='y', labelsize=FONTSIZE)  # Replace FONTSIZE with your desired size if needed
    ax.grid(True, axis='y')

plt.tight_layout()
#plt.savefig(f'./{output_directory}/Coefficient_growth_correlation_multi_v.png', bbox_inches = 'tight')

plt.show()
```

## Growth mixing considering only the other two strains, distinct from the local one

Here I weight the mixing by the area occupied by two species around the first one. Like an effective mixing.

```python
def shannon_entropy(proportions):
    # proportions: numpy array or pandas Series of relative abundances summing to <=1
    proportions = proportions[proportions > 0]  # remove zero to avoid log(0)
    return -np.sum(proportions * np.log(proportions))


# Your species
species_list = [0.0, 1.0, 2.0]

# Store correlation results
results = []
all_columns = list(df_combined.columns)
# get radii
# Pattern to extract the number after '_r_'
pattern = r'_r_(\d+)'

# Extract radius values from columns matching the pattern
radii = sorted({int(re.search(pattern, col).group(1)) 
                for col in all_columns if re.search(pattern, col)})

df_combined_d = df_combined.sample(frac=0.05, random_state=42)

for radius in radii:
    #plt.figure()
    for focal_species in species_list:
        # Filter cells of the focal species
        df_focal = df_combined_d[df_combined_d['Species'] == focal_species].copy()
        
        # Identify columns for the other two species
        other_species = [s for s in species_list if s != focal_species]
        occ_cols = [f'Occ_spec_{s}_r_{radius}' for s in other_species]
        occ_focal_col = [f'Occ_spec_{focal_species}_r_{radius}']
        
        # Compute "mixing" as the shannon entropy 
        
        df_focal['sum_other'] = df_focal[occ_cols].sum(axis=1)
        
        # Normalize occupations by sum to get proportions (avoid division by zero)
        df_focal['shann_mix_rescaled_single'] = df_focal.apply(lambda row: abs(shannon_entropy(row[occ_cols].fillna(0.0)/row['sum_other']))*row['sum_other'] if row['sum_other'] > 0 else 0.0,
        axis=1
        )
        # Get elongation rate
        df_focal = df_focal[df_focal['shann_mix_rescaled_single'] >0]
        growth = df_focal['Elongation_rate']
        mixing = df_focal['shann_mix_rescaled_single']
        
        #plt.scatter(growth, mixing, color = code_colors[focal_species], label = f'{code_name_species_mapping[focal_species]}')
        #sns.regplot(data=df_focal, x='Elongation_rate', y='shann_mix_rescaled_single', scatter=False, color=code_colors[focal_species], line_kws={"lw": 2, "ls": "--"})
        #plt.ylabel(f'Shannon Mixing Rescaled {radius}')
        # Compute correlation
        if len(growth) > 0 and len(mixing) > 0:
            corr, p_value = spearmanr(growth,
                                 mixing)
        else:
            corr = 0.0

        # Save result
        results.append({
            'radius': radius,
            'radius [um]': radius*ConversionFactor,
            'focal_species': focal_species,
            'Species_names': code_name_species_mapping[focal_species],
            'correlation': corr
        })
    #plt.show()
# Convert to DataFrame
df_corr = pd.DataFrame(results)

```

```python

# Prepare figure for bar plots: one subplot per species
fig, axes = plt.subplots(1, 3, figsize=(15, 5))

for i, (species_code, species_name) in enumerate(code_name_species_mapping.items()):
    ax = axes[i]
    
    species_closeby_1 = (species_code + 1) % 3
    species_closeby_2 = (species_code + 2) % 3
    color_ = code_colors[species_code]
    conc_species_1 = []
    conc_species_2 = []
    growth_species_focal = []

    # Collect data from all datasets
    for key, df_mixing in df_dict_growth_mixing.items():
        df_species = df_mixing[df_mixing['Species'] == species_code]
        conc_species_1.append(df_species[f'Occ_spec_{species_closeby_1}.0_r_153'].values)
        conc_species_2.append(df_species[f'Occ_spec_{species_closeby_2}.0_r_153'].values)
        growth_species_focal.append(df_species['Elongation_rate'].values)

    x1 = np.concatenate(conc_species_1)
    x2 = np.concatenate(conc_species_2)
    y = np.concatenate(growth_species_focal)

    # Handle NaNs and infinities
    x1 = np.nan_to_num(x1, nan=0.0, posinf=0.0, neginf=0.0)
    x2 = np.nan_to_num(x2, nan=0.0, posinf=0.0, neginf=0.0)

    # Add interaction term
    interaction = x1 * x2

    # Prepare design matrix with intercept
    X = np.column_stack((x1, x2, interaction))
    X = sm.add_constant(X)
    
    # Fit linear regression model
    model = sm.OLS(y, X).fit()

    # Extract coefficients and errors
    coef_names = [
        'Intercept',
        f'{code_name_species_mapping[species_closeby_1]}',
        f'{code_name_species_mapping[species_closeby_2]}',
        'Interaction'
    ]
    coefs = model.params
    errors = model.bse
    
    # Bar plot of coefficients with error bars
    ax.bar(coef_names, coefs, yerr=errors, capsize=5, color=color_)
    ax.set_title(f'{species_name}', fontsize=FONTSIZE)
    ax.set_ylabel('Coefficient', fontsize=FONTSIZE)
    ax.set_ylim(min(np.min(coefs) * 1.2, 0), max(np.max(coefs) * 1.2, 0))
    ax.tick_params(axis='x', labelsize=FONTSIZE)
    ax.tick_params(axis='y', labelsize=FONTSIZE)
    ax.grid(True, axis='y')

plt.tight_layout()
##plt.savefig(f'./{output_directory}/Coefficient_growth_correlation_with_interaction.png', bbox_inches='tight')
plt.show()
```

# Simulate with coefficients ahead

```python
from scipy.integrate import odeint

# Parameters
K = 1000  # Total carrying capacity
r = np.array([1.3, 1.3, 1.3])  # Growth rates for species 1, 2, 3

# Interaction matrix (including self-competition)
# Rows = species affected, Columns = species exerting influence
# Species 1 : PRO -ISO
# Species 2 : MET - ISO
# Species 3 : MET-PRO
alpha = np.array([
    [1.0, -0.15, -.45],  # Species 1 affected by (1,2,3)
    [0.05, 1.0, 0.13],  # Species 2 affected by (1,2,3)
    [-0.5, -0.5, 1.0]   # Species 3 affected by (1,2,3)
])

color_map = ['red', 'gold', 'blue']
# ODE system (no dilution since we now handle excess manually)
def dNdt(N, t, r, alpha, K):
    dN = np.zeros(3)
    for i in range(3):
        interaction_sum = sum(alpha[i, j] * N[j] for j in range(3))
        dN[i] = r[i] * N[i] * (1 - interaction_sum / K)
    return dN

# Apply stochastic truncation if total > K
def truncate_population(N, K):
    total = int(round(np.sum(N)))
    if total <= K:
        return N
    else:
        excess = total - K
        cell_list = []
        for i in range(3):
            cell_list.extend([i] * int(round(N[i])))
        np.random.shuffle(cell_list)
        for i in cell_list[:excess]:
            N[i] -= 1
        return np.clip(N, 0, None)

# Simulation parameters
N0 = np.array([100, 100, 100], dtype=float)
t = np.linspace(0, 300, 3000)
dt = t[1] - t[0]

# Integration using Euler method to allow stochastic truncation
pop_history = np.zeros((len(t), 3))
N = N0.copy()

for i, ti in enumerate(t):
    pop_history[i] = N
    dN = dNdt(N, ti, r, alpha, K)
    N = N + dN * dt
    N = truncate_population(N, K)

# Plotting
plt.figure(figsize=(10, 6))
for i in range(3):
    plt.plot(t, pop_history[:, i]/np.sum(pop_history, axis = 1), label=f'Species {i+1}', color = color_map[i])
plt.title('Three-Species Dynamics with Random Wash-Out Above 1000')
plt.xlabel('Time')
plt.ylabel('Population Size')
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()
```
